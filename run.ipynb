```python
from pathlib import Path
import subprocess
import sys

script_name = Path("dataset/create_rf3dpose_10gb_subset.py")
project_root = next(
    (p for p in (Path.cwd(), *Path.cwd().parents) if (p / script_name).is_file()),
    None,
)
if project_root is None:
    raise FileNotFoundError(
        "Open this notebook from the M4Human project, where dataset/ exists."
    )

# Source is the processed cache parent; it contains rf3dpose_all/*.lmdb.
source = project_root / "rf3dpose_all"
destination = project_root / "rf3dpose_10gb"
# Select whole subjects and/or action IDs; None means include all.
# For example: selected_people = ["p1", "p3"]
#              selected_actions = [2, 7]
selected_people = ["p1", "p3", "p4", "p5"]
selected_actions = None

command = [
    sys.executable,
    str(project_root / script_name),
    "--source",
    str(source),
    "--destination",
    str(destination),
    "--target-gib",
    "10",
    "--overwrite",
]
if selected_people:
    command.extend(["--select-people", *map(str, selected_people)])
if selected_actions:
    command.extend(["--select-actions", *map(str, selected_actions)])

subprocess.run(command, check=True)
```
